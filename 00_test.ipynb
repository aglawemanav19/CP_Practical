```python
#include <stdio.h>
int main()
{
    int a[5][5],b[5][5],c[5][5];
    int i,j,k;
    int r1,c1,r2,c2;
    printf("enter order of first matrix");
    scanf("%d%d",&r1,&c1);
    printf("enter order of second matrix");
    scanf("%d%d",&r2,&c2);
    if (c1==r2)
    {
        printf("\nenter %d elements of first matrix",r1*c1);
        for(i=0;i<r1;i++)
        {
            for(j=0;j<c1;j++)
            {
                scanf("%d",&a[i][j]);
            }
        }
        printf("\nenter %d elements of second matrix",r2*c2);
        for(i=0;i<r2;i++)
        {
            for(j=0;j<c2;j++)
            {
                scanf("%d",&b[i][j]);
            }
        }

        printf("\nfirst matrix is\n");
        for(i=0;i<r1;i++)
        {
            for(j=0;j<c1;j++)
            {
                printf("%d\t",a[i][j]);
            }
            printf("\n");
        }

        printf("\nsecond matrix is\n");
        for(i=0;i<r2;i++)
        {
            for(j=0;j<c2;j++)
            {
                printf("%d\t",b[i][j]);
            }
            printf("\n");
        }

        printf("\nproduct matrix is\n");
        for(i=0;i<r1;i++)
        {
            for(j=0;j<c2;j++)
            {
                c[i][j]=0;
                for(k=0;k<c1;k++)
                {
                    c[i][j]+=a[i][k]*b[k][j];
                }
                printf("%d\t",c[i][j]);
            }
            printf("\n");
        }

    }
    else
    {
        printf("matrix multiplication is not possible");
    }
}
```
